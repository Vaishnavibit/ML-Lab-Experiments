```python
#Z-Score — Outlier Detection
```

```python
from scipy.stats import zscore

# Calculate Z-score for numerical columns
numeric_cols = titanic.select_dtypes(include=['int64', 'float64']).columns

z_scores = titanic[numeric_cols].apply(zscore)

# Find rows having Z-score greater than 3 or less than -3
outliers = titanic[(z_scores.abs() > 3).any(axis=1)]

print("Number of outliers:", len(outliers))
print(outliers)
```

```python
from scipy.stats import zscore
import matplotlib.pyplot as plt
import seaborn as sns

# Calculate Z-score
numeric_cols = titanic.select_dtypes(include=['int64', 'float64']).columns
z_scores = titanic[numeric_cols].apply(zscore)

# Display Z-score
print("Z-Scores:")
print(z_scores.head())

# Boxplot for detecting outliers
plt.figure(figsize=(8,5))
sns.boxplot(data=titanic[numeric_cols])
plt.title("Boxplot for Outlier Detection")
plt.xlabel("Numerical Features")
plt.ylabel("Values")
plt.show()
```

```python
#Skewness — Data Distribution Chec
```

```python
import matplotlib.pyplot as plt

# Calculate skewness
skewness = titanic[numeric_cols].skew()

print("Skewness:")
print(skewness)

# Histogram
titanic[numeric_cols].hist(figsize=(10,6), bins=20)

plt.suptitle("Distribution of Numerical Features")
plt.show()
```

```python
#Variance — Measure of Data Spread
```

```python
# Calculate variance
variance = titanic[numeric_cols].var()

print("Variance of numerical columns:")
print(variance)
```

```python
#Standard Deviation
```

```python
# Calculate standard deviation
std_dev = titanic[numeric_cols].std()

print("Standard Deviation:")
print(std_dev)
```

```python
mean_values = titanic[numeric_cols].mean()
median_values = titanic[numeric_cols].median()

comparison = pd.DataFrame({
    'Mean': mean_values,
    'Median': median_values
})

print(comparison)

comparison.plot(kind='bar', figsize=(8,5))
plt.title("Mean vs Median")
plt.ylabel("Value")
plt.xticks(rotation=0)
plt.show()
```

```python
#Quantile / IQR — Another Outlier Check
```

```python
# Calculate Q1, Q3 and IQR
Q1 = titanic[numeric_cols].quantile(0.25)
Q3 = titanic[numeric_cols].quantile(0.75)

IQR = Q3 - Q1

print("Q1:")
print(Q1)

print("\nQ3:")
print(Q3)

print("\nIQR:")
print(IQR)
```
